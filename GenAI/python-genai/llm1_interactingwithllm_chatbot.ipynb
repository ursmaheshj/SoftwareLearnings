Running chatbot session with predefined context

```python
from google.genai import types
```

```python
config = types.GenerateContentConfig(
    system_instruction="Behave like, You are my mother who is strict and loving at the same time" \
    "who taunts and scold me for bad things and appreciate for good things, she never agrees with me" \
    "she sometimes makes me feel like a friend and sometimes becomes very strict person. i am her cute, sweet girl so make it like she is talking to girl" \
    "she calls me niki,nikke as my name is nikita. i call her aai in marathi, she knows marathi only and a little bit of english" \
    "dont break the character make it seem like a real person talking in marathi, usually she tells me stories from my village and" \
    "i tell her what i do in pune what happened with roommates and all, she talks in 6,7 word sentences",
    temperature=0.7,
    top_p=0.95,
    max_output_tokens=1024
)
```

```python
chat = client.chats.create(
    model="gemini-2.5-flash-lite",
    config=config,
    history=[]
)
```

```python
while True:
    message = input("Enter message")
    if message.lower()=="quit":
        break
    response = chat.send_message(message)
    print(f"aai: {response.text}")
```
